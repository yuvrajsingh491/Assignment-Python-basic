Q 6 Why are lists mutable, but tuples are immutable

 - Why lists are mutable:

 Designed for changeable data: Lists are meant to be flexible — you can add, remove, or update items.

 Implementation: Lists are backed by dynamic arrays in memory. Python allows modifying their contents directly (e.g., appending or reassigning elements).

 Use Case: Ideal when you need a collection that will be updated during runtime.

  Why tuples are immutable:
Designed for fixed data: Tuples are meant to represent constant collections — like coordinates, RGB values, or database keys.

 Implementation: Once a tuple is created, it’s locked in memory. Python does this intentionally for:

 Efficiency (they’re faster than lists)

 Hashability (they can be used as dictionary keys or set elements)

 Use Case: Use tuples when you want to ensure the data remains unchanged.

 Q 7 What is the difference between “==” and “is” operators in Python

    - The == and is operators in Python may seem similar but they check very different things:

    == (Equality Operator)
  Checks if two variables have the same value, regardless of whether they are the same object in memory.

  Compares contents.
 a = [1, 2, 3]
 b = [1, 2, 3]
print(a == b)   # ✅ True — same contents
   
   is (Identity Operator)
   Checks if two variables refer to the exact same object in memory.

   Compares object identity (i.e. memory address).

   print(a is b)   # ❌ False — different objects with the same content

 Q 10  What is the difference between implicit and explicit type casting
  
  -The difference between implicit and explicit type casting in Python lies in who performs the conversion—Python or the programmer—and when it happens.

   1. Implicit Type Casting (Automatic by Python)
 Python automatically converts one data type to another when it’s safe and won’t lose data.

 Common in operations involving mixed numeric types.
  
  x = 10       # int
y = 3.5      # float
z = x + y    # Python implicitly converts int to float
print(z)     # 13.5
print(type(z))  # <class 'float'>

  2. Explicit Type Casting (Manual by the Programmer)
  You manually convert the type using built-in functions.

  Required when types are incompatible or when you want precise control.

  x = "100"
y = int(x)    # Explicitly convert str to int
print(y + 5)  # 105

Q12 How does the elif statement work
   
- The elif (short for "else if") statement in Python is used to check multiple conditions after an initial if condition.

   How elif Works:
   Python checks the first if condition.

  If it's True, the corresponding block runs, and the rest of the elif/else blocks are skipped.

  If the if is False, it checks the elif condition.

  If an elif is True, its block runs, and Python skips the rest.

  If none of the if or elif conditions are True, the else block (if provided) runs.
  
  x = 10

if x > 20:
    print("Greater than 20")
elif x > 15:
    print("Greater than 15 but not more than 20")
elif x > 5:
    print("Greater than 5 but not more than 15")
else:
    print("5 or less")

   for Loop – Used when you know ahead of time how many times to repeat

Great for counted loops or looping through items.
  
  for i in range(5):

Q7  Write a program that performs basic arithmetic operations on two user-input numbers.

```python
# Get input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Handle division safely
if num2 != 0:
    division = num1 / num2
else:
    division = "Undefined (cannot divide by zero)"

# Display the results
print("\nArithmetic Operations:")
print(f"{num1} + {num2} = {addition}")
print(f"{num1} - {num2} = {subtraction}")
print(f"{num1} * {num2} = {multiplication}")
print(f"{num1} / {num2} = {division}")
```

Q9  Write a Python program to convert user input from string to integer, float, and boolean types.

```python
# Get user input as a string
user_input = input("Enter a value: ")

# Convert to integer
try:
    int_value = int(user_input)
    print(f"Integer conversion: {int_value}")
except ValueError:
    print("Cannot convert to integer.")

# Convert to float
try:
    float_value = float(user_input)
    print(f"Float conversion: {float_value}")
except ValueError:
    print("Cannot convert to float.")

# Convert to boolean
# Any non-empty string (except "False", "false", "0", etc.) is considered True
bool_value = user_input.strip().lower() not in ["", "false", "0", "none"]
print(f"Boolean conversion: {bool_value}")
```

Q10 Write code to demonstrate type casting with list elements

```python
# List of string numbers
str_list = ["10", "20", "30", "0", ""]

print("Original list (strings):", str_list)

# Convert to integers
try:
    int_list = [int(item) for item in str_list]
    print("Converted to integers:", int_list)
except ValueError as e:
    print("Error converting to integers:", e)

# Convert to floats
try:
    float_list = [float(item) for item in str_list]
    print("Converted to floats:", float_list)
except ValueError as e:
    print("Error converting to floats:", e)

# Convert to booleans
bool_list = [bool(item) for item in str_list]
print("Converted to booleans:", bool_list)
```

Q13 Write a Python program to find the sum of all even numbers between 1 and 50.



```python
# Initialize sum
even_sum = 0

# Loop through numbers 1 to 50
for num in range(1, 51):
    if num % 2 == 0:  # Check if the number is even
        even_sum += num

# Display the result
print("Sum of even numbers from 1 to 50 is:", even_sum)
```

Q14 Write a program to reverse a string using a while loop.

```python
# Get input from the user
original = input("Enter a string: ")

# Initialize variables
reversed_str = ""
index = len(original) - 1

# Use while loop to reverse the string
while index >= 0:
    reversed_str += original[index]
    index -= 1

# Display the reversed string
print("Reversed string:", reversed_str)
```

Q15 Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
# Get input from the user
num = int(input("Enter a non-negative integer: "))

# Check for negative input
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1
    i = 1

    while i <= num:
        factorial *= i
        i += 1

    print(f"The factorial of {num} is: {factorial}")
```
